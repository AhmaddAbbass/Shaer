```python
!pip install -q datasets huggingface_hub

from datasets import load_dataset, Dataset
from huggingface_hub import login

# 👉 Login
login()

############################################################
# 🔥 1) RAM-SAFE EXPANSION USING GENERATOR
############################################################

def poem_expansion_generator(ds):
    for example in ds:
        verses = example.get("poem_verses", None)

        if not isinstance(verses, list):
            continue

        if len(verses) % 2 != 0:
            continue  # skip odd rows just in case

        num_verses = len(verses) // 2

        for seq in range(num_verses):
            start = 2 * seq
            end = start + 2

            yield {
                **{k: v for k, v in example.items() if k != "poem_verses"},
                "target_verse": verses[start:end],
                "previous_verses": verses[:start],
                "sequence_number": seq + 1
            }

def expand_poem_dataset_streaming(ds):
    return Dataset.from_generator(lambda: poem_expansion_generator(ds))

############################################################
# 🔥 2) EXPAND TRAINING SPLIT
############################################################

train_ds = load_dataset("Shaer-AI/ashaar-training", split="train")
print("Original train rows:", train_ds.num_rows)

train_expanded = expand_poem_dataset_streaming(train_ds)
print("Expanded train rows:", train_expanded.num_rows)

train_expanded.push_to_hub("Shaer-AI/ashaar-training_split")
print("✅ Uploaded training split")

############################################################
# 🔥 3) EXPAND TESTING SPLIT
############################################################

test_ds = load_dataset("Shaer-AI/ashaar-testing", split="train")
print("Original test rows:", test_ds.num_rows)

test_expanded = expand_poem_dataset_streaming(test_ds)
print("Expanded test rows:", test_expanded.num_rows)

test_expanded.push_to_hub("Shaer-AI/ashaar-testing_split")
print("✅ Uploaded testing split")

############################################################
# 🔥 4) EXPAND VALIDATION SPLIT
############################################################

valid_ds = load_dataset("Shaer-AI/ashaar-validation", split="train")
print("Original validation rows:", valid_ds.num_rows)

valid_expanded = expand_poem_dataset_streaming(valid_ds)
print("Expanded validation rows:", valid_expanded.num_rows)

valid_expanded.push_to_hub("Shaer-AI/ashaar-validation_split")
print("✅ Uploaded validation split")

print("\n🎉 ALL SPLITS EXPANDED + UPLOADED WITH ZERO RAM CRASH!")
```
